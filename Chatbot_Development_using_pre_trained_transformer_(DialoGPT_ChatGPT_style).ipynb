```python
!pip install -q transformers torch

from transformers import pipeline

chatbot = pipeline("text-generation", model="microsoft/DialoGPT-small")

text = "Hello! How are you?"

result = chatbot(
    text,
    max_new_tokens=40,
    pad_token_id=50256,
    do_sample=True,
    top_k=50,
    top_p=0.95
)

print("User:", text)
print("Chatbot:", result[0]["generated_text"][len(text):])
```
