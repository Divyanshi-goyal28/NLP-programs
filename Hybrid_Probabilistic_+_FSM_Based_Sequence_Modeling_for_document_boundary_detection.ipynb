```python
import re

text = """
This is the first paragraph of the document.

This is the second paragraph of the document.

References
1. Smith, A. Research in NLP.
2. John, B. Machine Learning.
"""

sentences = re.split(r'\n\s*\n', text.strip())

state = "BODY"

for sentence in sentences:
    sentence = sentence.strip()

    if not sentence:
        continue

    probability = 0.9 if sentence.lower().startswith("references") else 0.1

    if probability > 0.5:
        state = "REFERENCES"
    elif state != "REFERENCES":
        state = "BODY"

    print("State:", state)
    print("Text:", sentence)
    print()
```
