```python

import spacy

nlp = spacy.load("en_core_web_sm")

text = "Rahul is a student. He studies NLP. Rahul likes it."

doc = nlp(text)

print("Original Text:")
print(text)

print("\nCoreference Resolution:")

for token in doc:
    if token.pos_ == "PRON":
        print(token.text, "-> Pronoun referring to an entity in the text")
```
