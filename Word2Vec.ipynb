```python
!pip install gensim

from gensim.models import Word2Vec

corpus = [
    ["natural", "language", "processing", "is", "interesting"],
    ["natural", "language", "processing", "uses", "machine", "learning"],
    ["machine", "learning", "is", "a", "part", "of", "artificial", "intelligence"],
    ["deep", "learning", "is", "used", "in", "natural", "language", "processing"],
    ["artificial", "intelligence", "and", "machine", "learning", "are", "powerful"],
    ["python", "is", "widely", "used", "for", "machine", "learning"],
    ["python", "is", "also", "used", "for", "natural", "language", "processing"]
]

model = Word2Vec(
    sentences=corpus,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=1
)


print("Vocabulary:")
print(list(model.wv.index_to_key))


word = "machine"

print("\nWord Vector for:", word)
print(model.wv[word])


print("\nWords similar to 'machine':")
print(model.wv.most_similar("machine", topn=5))


print("\nSimilarity between 'machine' and 'learning':")
print(model.wv.similarity("machine", "learning"))

model.save("custom_word2vec.model")

print("\nWord2Vec model trained and saved successfully.")
```
