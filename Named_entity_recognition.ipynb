```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Elon Musk founded SpaceX in California in 2002. Microsoft is based in the United States."

doc = nlp(text)

print("Text:")
print(text)

print("\nNamed Entities:")

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```
