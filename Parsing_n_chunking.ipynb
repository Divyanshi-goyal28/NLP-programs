```python
import re
import spacy

nlp = spacy.load("en_core_web_sm")

text = "The young student is reading a book in the library."

doc = nlp(text)

print("Sentence:")
print(text)

print("\nParsing:")
for token in doc:
    print(token.text, "->", token.pos_, "->", token.dep_)

pattern = r"\b(?:the|a|an)\s+\w+(?:\s+\w+)*"
chunks = re.findall(pattern, text, re.IGNORECASE)

print("\nRegEx Chunking:")
for chunk in chunks:
    print(chunk)

print("\nspaCy Noun Chunks:")
for chunk in doc.noun_chunks:
    print(chunk.text)
```
