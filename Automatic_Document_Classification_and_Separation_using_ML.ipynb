```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

documents = [
    "The team won the football match",
    "The player scored a goal",
    "Python is a programming language",
    "Machine learning uses algorithms"
]

labels = ["Sports", "Sports", "Technology", "Technology"]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(documents)

model = MultinomialNB()
model.fit(X, labels)

new_documents = [
    "The team scored a goal",
    "Python programming is useful"
]

predictions = model.predict(vectorizer.transform(new_documents))

for document, label in zip(new_documents, predictions):
    print("Document:", document)
    print("Category:", label)
    print()
```
