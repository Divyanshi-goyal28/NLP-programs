```python
!pip install -q gensim POT

import gensim.downloader as api

model = api.load("glove-wiki-gigaword-50")

text1 = "I like apples"
text2 = "I like oranges"

distance = model.wmdistance(text1.split(), text2.split())

print("Text 1:", text1)
print("Text 2:", text2)
print("Word Mover's Distance:", distance)
```
