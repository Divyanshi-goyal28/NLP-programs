```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Rahul works at Google in Delhi. He joined the company in 2023."

doc = nlp(text)

print("Unstructured Text:")
print(text)

print("\nExtracted Information:")

for ent in doc.ents:
    print(ent.text, ":", ent.label_)
```
